# configure the root logger

```python
import logging
logging.basicConfig(level= logging.WARNING)

logging.debug("this is debug message")
logging.info("this is info message")
logging.warning("this is warning message")
logging.error("this is error message")
logging.critical("this is critical message")
```

```python
import logging
logging.basicConfig(level= logging.DEBUG, force=True)

logging.debug("this is debug message")
logging.info("this is info message")
logging.warning("this is warning message")
logging.error("this is error message")
logging.critical("this is critical message")
```

Format the log messages

```python
import logging
logging.basicConfig(level= logging.DEBUG, force=True, format="%(asctime)s - %(levelname)s - %(message)s")

logging.debug("this is debug message")
logging.info("this is info message")
logging.warning("this is warning message")
logging.error("this is error message")
logging.critical("this is critical message")
```

Persist Logs


```python
import logging
logging.basicConfig(level= logging.DEBUG, force=True, format="%(asctime)s - %(levelname)s - %(message)s",
                    filename="app.log")

logging.debug("this is debug message")
logging.info("this is info message")
logging.warning("this is warning message")
logging.error("this is error message")
logging.critical("this is critical message")
```

Handlers are object that decide where to0 send log messages

```python
import logging

# create handlers
console_handler = logging.StreamHandler()
file_handler = logging.FileHandler("app.log")

logging.basicConfig(level= logging.DEBUG, force=True, format="%(asctime)s - %(levelname)s - %(message)s",
                    handlers = [console_handler, file_handler])

logging.debug("this is debug message")
logging.info("this is info message")
logging.warning("this is warning message")
logging.error("this is error message")
logging.critical("this is critical message")
```

We need child logger to know from which part of app the log is comming from 

```python
import logging

# root logger
logging.basicConfig(level= logging.INFO, force=True, format="%(asctime)s - %(levelname)s - %(message)s")

# create a child logger
logger=logging.getLogger(__name__)

logger.info("Application Started")
logger.info("application finished")
```
